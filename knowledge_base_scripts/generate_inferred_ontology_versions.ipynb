```python
from rdflib import Graph
import owlrl

def run_inference(input_path: str, output_path: str):
    schema = Graph()
    schema.parse(input_path)
    print(f"Loaded ontology with {len(schema)} triples")

    owlrl.DeductiveClosure(owlrl.OWLRL_Semantics).expand(schema)
    print("Finished reasoning")

    print("Writing inferred model..")
    schema.serialize(destination=output_path, format="xml")
    print("done")

if __name__ == "__main__":
    INPUT  = "../ontology/COMON.owl"
    OUTPUT = "../ontology/COMON-inferred.owl"
    run_inference(INPUT, OUTPUT)
```
